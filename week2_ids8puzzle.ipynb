```python
class PuzzleNode:
    def __init__(self, state, parent=None, move=None):
        self.state = state
        self.parent = parent
        self.move = move

    def __eq__(self, other):
        return self.state == other.state

    def __hash__(self):
        return hash(str(self.state))


def print_state(state):
    for row in state:
        print(" ".join(str(x) for x in row))
    print()


def find_empty(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j


def get_actions(state):
    moves = []
    row, col = find_empty(state)

    if row > 0:
        moves.append("up")
    if row < 2:
        moves.append("down")
    if col > 0:
        moves.append("left")
    if col < 2:
        moves.append("right")

    return moves


def make_move(state, move):
    new_state = [row[:] for row in state]
    row, col = find_empty(new_state)

    if move == "up":
        new_state[row][col], new_state[row - 1][col] = \
            new_state[row - 1][col], new_state[row][col]

    elif move == "down":
        new_state[row][col], new_state[row + 1][col] = \
            new_state[row + 1][col], new_state[row][col]

    elif move == "left":
        new_state[row][col], new_state[row][col - 1] = \
            new_state[row][col - 1], new_state[row][col]

    elif move == "right":
        new_state[row][col], new_state[row][col + 1] = \
            new_state[row][col + 1], new_state[row][col]

    return new_state


def depth_limited_search(start, goal, limit):
    stack = [(PuzzleNode(start), 0)]
    path_nodes = set()

    while stack:
        current, depth = stack.pop()

        if current.state == goal:
            solution = []
            node = current

            while node.parent is not None:
                solution.append(node.move)
                node = node.parent

            solution.reverse()
            return solution

        if depth < limit:
            path_nodes.add(current)

            for move in get_actions(current.state):
                next_state = make_move(current.state, move)
                child = PuzzleNode(next_state, current, move)

                if child not in path_nodes:
                    stack.append((child, depth + 1))

    return None


def iterative_deepening_search(start, goal):
    depth = 0

    while True:
        print("Searching at depth limit:", depth)

        result = depth_limited_search(start, goal, depth)

        if result is not None:
            print("\nSolution found!")
            print("Depth:", depth)
            print("Moves:", " -> ".join(result))
            return result

        depth += 1


def show_solution(start, moves):
    current_state = start

    print("\n--- Solution Path ---")
    print("Initial State:")
    print_state(current_state)

    for i, move in enumerate(moves, 1):
        current_state = make_move(current_state, move)

        print("Move", i, ":", move)
        print_state(current_state)


if __name__ == "__main__":

    start_state = [
        [2, 8, 3],
        [1, 6, 4],
        [7, 0, 5]
    ]

    goal_state = [
        [1, 2, 3],
        [8, 0, 4],
        [7, 6, 5]
    ]
    print("1BM24CS147")
    print("Initial State:")
    print_state(start_state)

    print("Goal State:")
    print_state(goal_state)

    print("Starting Iterative Deepening Search...\n")

    solution = iterative_deepening_search(start_state, goal_state)

    show_solution(start_state, solution)
```
