```python
import heapq

def parse_state(prompt, default_val):
    try:
        user_input = input(f"{prompt}: ").strip()
        if not user_input:
            return default_val
        state = tuple(map(int, user_input.split()))
        if len(state) != 9 or set(state) != set(range(9)):
            return default_val
        return state
    except Exception:
        return default_val

initial = parse_state("Enter Initial State", (2, 8, 3, 1, 6, 4, 7, 0, 5))
goal = parse_state("Enter Goal State", (1, 2, 3, 8, 0, 4, 7, 6, 5))

def misplaced(s):
    return sum(s[i] != goal[i] and s[i] != 0 for i in range(9))

def neighbors(s):
    z = s.index(0)
    r, c = divmod(z, 3)
    for dr, dc, move in [(-1, 0, "Up"), (1, 0, "Down"), (0, -1, "Left"), (0, 1, "Right")]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            p = nr * 3 + nc
            t = list(s)
            t[z], t[p] = t[p], t[z]
            yield move, tuple(t)

def a_star(h):
    pq = [(h(initial), 0, initial, [])]
    visited = set()
    while pq:
        f, g, s, path = heapq.heappop(pq)
        if s in visited: continue
        visited.add(s)
        if s == goal: return path
        for move, ns in neighbors(s):
            if ns not in visited:
                ng = g + 1
                nh = h(ns)
                heapq.heappush(pq, (ng + nh, ng, ns, path + [(move, ns, ng, nh, ng + nh)]))
    return None

def print_puzzle(s):
    res = []
    for i in range(0, 9, 3):
        res.append(" ".join("_" if x == 0 else str(x) for x in s[i:i+3]))
    return "\n".join(res)

print("Name:MADHUSUDHAN M G")
print("USN: 1BM24CS154")
print("-" * 20)

path = a_star(misplaced)
output = []
output.append("Initial State:\n" + print_puzzle(initial) + "\n")

if path is None:
    output.append("No solution found.")
else:
    for step, (move, state, g, hn, f) in enumerate(path, 1):
        output.append(f"Step: {step}")
        output.append(f"Move: {move}")
        output.append(f"g(n) = {g}  h(n) = {hn}  f(n) = {f}")
        output.append(print_puzzle(state) + "\n")
    output.append(f"Total moves: {len(path)}")

print("\n".join(output))
```
